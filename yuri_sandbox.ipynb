```python
from pathlib import Path
import pandas as pd

data_dir = Path("data/")  # folder containing 2005.jsonl … 2025.jsonl

df = pd.concat(
    (
        pd.read_json(f, lines=True).assign(year=int(f.stem))
        for f in sorted(data_dir.glob("*.jsonl"))
    ),
    ignore_index=True,
)
```

```python
print(df.shape)
print(df.info())
```

```python
print(df["regionId"].value_counts().sort_index())
```

```python
import pandas as pd

# regionIds (plural, a list) may differ from regionId
df["n_regionIds"] = df["regionIds"].apply(lambda x: len(x) if isinstance(x, list) else 0)
print((df["n_regionIds"] > 0).mean())

# geotags: list of dicts, e.g. [{"tag": "Sachsen"}]
geo = df["geotags"].explode().dropna().apply(lambda d: d.get("tag") if isinstance(d, dict) else d)
print(geo.value_counts().head(40))

# editor tags often contain state names ("Sachsen", "Landtagswahl Thüringen")
tags = df["tags"].explode().dropna().apply(lambda d: d.get("tag") if isinstance(d, dict) else d)
print(tags[tags.str.contains("Sachsen|Thüringen|Brandenburg|Mecklenburg|Bayern|Hessen|Saarland", na=False)].value_counts().head(40))
```
